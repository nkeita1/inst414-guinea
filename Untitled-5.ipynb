```python
%matplotlib inline
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.cluster import KMeans
```

```python
wine = pd.read_csv("winequality-red.csv", sep=";")
print(wine.shape)
wine.head()
```

```python
print("Rows and columns:", wine.shape)
print("Missing values:", wine.isna().sum().sum())
print(wine.mean())
```

```python
features = []
for name in wine.columns:
    if name != "quality":
        features.append(name)

lab = wine[features]
scaled = (lab - lab.mean()) / lab.std()
print(scaled.shape)
```

```python
ks = []
inertias = []

for k in range(1, 13):
    model = KMeans(n_clusters=k, n_init=10, random_state=414)
    model.fit(scaled)
    ks.append(k)
    inertias.append(model.inertia_)
    print(k, model.inertia_)

print("Improvement from adding one more cluster:")
for i in range(1, len(inertias)):
    print("k =", ks[i], ":", inertias[i-1] - inertias[i])

plt.figure(figsize=(7, 4))
plt.plot(ks, inertias, marker="o")
plt.xlabel("Number of clusters (k)")
plt.ylabel("Inertia (lower = tighter clusters)")
plt.title("Elbow plot: red wine measurements")
plt.xticks(ks)
plt.show()
```

```python
k = 7
model = KMeans(n_clusters=k, n_init=10, random_state=414)
model.fit(scaled)
labels = model.predict(scaled)

wine["cluster"] = labels
print(wine["cluster"].value_counts())
```

```python
for seed in [1, 2, 3]:
    other = KMeans(n_clusters=7, n_init=10, random_state=seed).fit(scaled)
    print("random_state", seed, "inertia", other.inertia_)
```

```python
profile = wine.groupby("cluster").mean()
profile["wines"] = wine["cluster"].value_counts()
profile
```

```python
for cluster_id, group in wine.groupby("cluster"):
    print("Cluster", cluster_id, "- wines:", len(group))
    print(group[["alcohol", "volatile acidity", "fixed acidity", "total sulfur dioxide", "residual sugar", "chlorides", "sulphates", "quality"]].sample(2, random_state=0))
    print()
```
