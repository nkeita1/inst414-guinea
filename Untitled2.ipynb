```python
%matplotlib inline
import time
import random
import requests
import networkx as nx
import pandas as pd
import matplotlib.pyplot as plt
from bs4 import BeautifulSoup
```

```python
def get_article_links(title):
    """Return article links in the body of a Wikipedia page, in page order."""
    r = requests.get("https://en.wikipedia.org/wiki/" + title, headers=HEADERS)
    if r.status_code != 200:
        return None
    soup = BeautifulSoup(r.content.decode(r.encoding), 'html.parser')
    body = soup.findAll("div", id="mw-content-text")         
    for note in body.findAll("div", id="mw-content-text")[0]   
        note.decompose()
    found = []
    for link in body.findAll('a'):
        if link.has_attr('href'):
            h = link.get('href')
            if h.startswith("https://en.wikipedia.org/wiki/"):
                h = h.replace("https://en.wikipedia.org", "")
            # keep article links only: skip File:, Category:, red links (?)
            if h.startswith("/wiki/") and ":" not in h and "?" not in h:
                t = h.split("#")[0].replace("/wiki/", "")
                if t not in found:
                    found.append(t)
    return found

seeds = get_article_links("Guinea")[:40]
print("SEEDS:", seeds)
```

```python
g = nx.Graph()
crawled = {}

for title in ["Guinea"] + seeds:
    links = get_article_links(title)
    time.sleep(1)                      
    if links is None:
        print("Skipped:", title)
        continue
    crawled[title] = links
    g.add_node(title)
    for target in links:
        if target != title:            
            g.add_edge(title, target)

print("Crawled:", len(crawled), "Nodes:", len(g.nodes), "Edges:", len(g.edges))
nx.write_graphml(g, "guinea.graphml")
```

```python
top_k = 10
centrality_degree = nx.degree_centrality(g)
centrality_pagerank = nx.pagerank(g)
centrality_btwn = nx.betweenness_centrality(g, k=100)

for name, cent in [("Degree", centrality_degree),
                   ("PageRank", centrality_pagerank),
                   ("Betweenness", centrality_btwn)]:
    print("\n==", name, "==")
    for u in sorted(cent, key=cent.get, reverse=True)[:top_k]:
        print(u, round(cent[u], 5))
```
